```python
import matplotlib.pyplot as plt

numeric_cols = [
    "total_folios_crore",
    "equity_folios_crore",
    "debt_folios_crore",
    "hybrid_folios_crore",
    "others_folios_crore"
]

for col in numeric_cols:
    plt.figure(figsize=(6,3))
    plt.boxplot(data[col])
    plt.title(col)
    plt.show()
```
