```python
print("Negative 1-Year Returns:", (df["return_1yr_pct"] < 0).sum())
print("Negative 3-Year Returns:", (df["return_3yr_pct"] < 0).sum())
print("Negative 5-Year Returns:", (df["return_5yr_pct"] < 0).sum())

print("AUM <= 0:", (df["aum_crore"] <= 0).sum())

print("Expense Ratio <= 0:", (df["expense_ratio_pct"] <= 0).sum())

print("Beta <= 0:", (df["beta"] <= 0).sum())

print("Morningstar Rating outside 1-5:",
      ((df["morningstar_rating"] < 1) |
       (df["morningstar_rating"] > 5)).sum())
```

```python
import matplotlib.pyplot as plt

numeric_cols = [
    "return_1yr_pct",
    "return_3yr_pct",
    "return_5yr_pct",
    "benchmark_3yr_pct",
    "alpha",
    "beta",
    "sharpe_ratio",
    "sortino_ratio",
    "std_dev_ann_pct",
    "max_drawdown_pct",
    "aum_crore",
    "expense_ratio_pct"
]

for col in numeric_cols:
    plt.figure(figsize=(6,3))
    plt.boxplot(df[col])
    plt.title(col)
    plt.show()
```

```python
numeric_cols = [
    "return_1yr_pct",
    "return_3yr_pct",
    "return_5yr_pct",
    "benchmark_3yr_pct",
    "alpha",
    "beta",
    "sharpe_ratio",
    "sortino_ratio",
    "std_dev_ann_pct",
    "max_drawdown_pct",
    "aum_crore",
    "expense_ratio_pct"
]

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"{col}: {len(outliers)} outliers")
```

```python
### Outlier Analysis

print("Outliers were identified using the IQR method across multiple performance metrics such as returns, benchmark returns, beta, Sharpe ratio, Sortino ratio, and annualized standard deviation.")

print("After reviewing the dataset, these outliers were determined to represent genuine differences in mutual fund performance and risk characteristics rather than data quality issues or entry errors.")  

print("Therefore, no outliers were removed.")
```
